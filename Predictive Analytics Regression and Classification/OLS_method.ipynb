```python
# Load Pandas and Numpy

import pandas as pd
import numpy as np

# Read CSV file 

df = pd.read_csv('mtcars.csv')
```

```python
df.head()
```

```python
# Sample Size
n = df.shape[0]
n
```

```python
# Target variable is mpg. We define it as matrix of order nX1.

y = np.transpose(np.asmatrix(df['mpg']))
y[:]
```

```python
# First we define the intercept of the model
Intercept = pd.DataFrame(np.ones(n))
Intercept
```

```python
# We consider 3 variables as predictor i.e. hp, weight and displacement as predictor for mpg
Xdf = df[['hp','wt','disp']]

# Add intercept to Xdf
Xdf = pd.concat([Intercept,Xdf],axis=1)
Xdf.head()
```

The design matrix X is defined as matrix x from Xdf which is of the datatype dataframe.

```python
X =  np.asmatrix(Xdf)
X[:,:]
```

```python
Xt = np.transpose(X)
```

Calculate X'X

```python
Xt*X
```

Calculate inverse of X'X

```python
XtX_inv = np.linalg.inv(Xt*X)
XtX_inv
```

Now calculating regressor beta cap = (X'X)^-1.X'y

```python
beta_hat = XtX_inv*Xt*y
beta_hat
```

```python
np.round(beta_hat,4)
```

Now we will implement ordinary least square using statsmodel

```python
import statsmodels.api as sm
```

```python
#fit linear regression model
model = sm.OLS(y,X).fit()

#view model summary
print(model.summary())
```

```python
from sklearn.linear_model import LinearRegression

X = np.asarray(X)
y = np.asarray(y)

model = LinearRegression(fit_intercept = False)
model.fit(X,y)
```

```python
beta = model.coef_
alpha = model.intercept_
```
