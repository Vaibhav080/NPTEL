```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
def MLE_theta(sample_points):
    # This code computes the MLE estimate of theta
    log_sum = 0
    for x in sample_points:
        log_sum = log_sum + math.log(x)
    n = len(sample_points)
    return n/log_sum
```

```python
def show_pdf(x, theta):
    # Distribution pdf
    return theta / x**(theta + 1)
```

```python
def main():
    sample_points = [2.5,3.1,4.0,2.8,3.7,8.9,4.3,10.3,1.1,5.6]
    theta = MLE_theta(sample_points)
    print(theta)

    # Plot the histogram of the sample data
    plt.hist(sample_points, bins = 20, density = True, color = 'lightblue', label = 'sample_data')

    #Generate the values of the fitted data
    x_values = np.linspace(min(sample_points), max(sample_points), 10000)
    y_values = [show_pdf(x, theta) for x in x_values]

    # Plot the fitted distribution

    plt.plot(x_values, y_values,color = 'red', label = 'Distribution Fit')

    plt.xlabel('Values')
    plt.ylabel('Probability Density')
    plt.title('Distribution fit to sample data')
    plt.legend()
    plt.show()
```

```python
if(__name__ == "__main__"):
    main()
```
