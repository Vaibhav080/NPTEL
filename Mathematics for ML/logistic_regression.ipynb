```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_curve, auc
```

```python
# Loading the dataset for classification

Data = load_breast_cancer()
data_features = Data.data # Extracting the features from the dataset
data_target = Data.target # Extracting the target label
Labels = Data.target
Feature_names = Data.feature_names
print("Feature_Names: ", Feature_names)

Num_Samples = data_features.shape[0]
print("\nnumber of samples: ", Num_Samples)
```

```python
# Print the number of features
Num_Features = data_features.shape[1]
print("Number of features: ", Num_Features)

# Print the number of classes
Num_classes = len(set(Labels))
print("Number of Classes: ", Num_classes)
```

```python
# Divinding the dataset into the training and testing part (80:20)
train_x, test_x, train_y, test_y = train_test_split(data_features, data_target, test_size = 0.2, random_state = 42)

# Normalizing the features
scaler = StandardScaler()
x_train = scaler.fit_transform(train_x)
x_test = scaler.transform(test_x)

# Number of features in the dataset
num_features = x_train.shape[1]
print("number of features in the dataset: ", num_features)
```

```python
# Training the considered logistic regression model for the considered dataset
Logis_Reg = LogisticRegression()
Logis_Reg.fit(x_train,train_y)
y_prediction = Logis_Reg.predict(x_test)
accuracy = accuracy_score(test_y , y_prediction)
print("\nAccuracy of the model: ",accuracy)
```
