```python
from numpy import arange
from numpy.random import rand
from matplotlib import pyplot
```

```python
#Objective function

def objective(x):
    return x**4-x**2+2*x-8
```

```python
#Range for imput
x_min,x_max = -10,10
#Sample Points
sample = x_min+rand(150)*(x_max-x_min)
#evaluate the sample
y= objective(sample)
```

```python
#locate the best solution
opt=0
for i in range(len(sample)):
    if y[i] < y[opt]:
        opt=i
#Optimal through random search
print('Optimal f(%.4f) = %.4f' % (sample[opt],y[opt]))
#sample input range creatd uniformly with 0.1 increments
inputs = arange(x_min,x_max,0.1)
#Tfunctiona values
y_value = objective(inputs)
#plot
pyplot.plot(inputs, y_value)
pyplot.scatter(sample,y)
pyplot.axvline(x=sample[opt], ls='--', color='green')
pyplot.show()
```
