```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from tabulate import tabulate
import os
```

```python
# Considered function to fit in the data

def fitting_function(x_value,w_weight,b_bias):
    return w_weight*np.sin(5*x_value) +b_bias
def cost_function(x_value,w_weight,b_bias,Y_true,Lambda):
    return np.mean((fitting_function(x_value, w_weight,b_bias)-Y_true)**2 + Lambda*np.sum(np.abs(w_weight)))
def computing_gradient_cost(x_value,w_weight,b_bias,Y_true,Lambda):
    Err = fitting_function(x_value,w_weight,b_bias) - Y_true # Calculating error in the gradient function
    sub_gradient_w_weight = 2*np.mean(Err*np.sin(5*x_value)) + Lambda * np.sign(w_weight) # Calculating the derivative wrt w_weight
    gradient_b_bias = 2*np.mean(Err) # Calculating the derivative with respect to b_bias
    return sub_gradient_w_weight,gradient_b_bias # returning weight and bias
```

```python
# Initializing the weight w and bias b
weight = 3.0 # initializing the weight
bias = 1.0 # initializing the bias
learning_rate = 0.1 # giving the learning rate
num_iter = 50 # number of iterations
Lambda = 0.3 # Lambda value

# Random data generation for X and target data (Y)

X_value = np.linspace(0,2*np.pi,100) #Random values for x
Y_true = np.linspace(0,2,100) # randomly generated values to ft the function

# Combine x and y values into table format
data_points = np.vstack((X_value[:10], Y_true[:10]))

#Print the table
print(tabulate(data_points,headers = ['X_value','Y_true'],tablefmt='grid'))
```

```python
# Initializing the random weights and biases and correspondng cost functions for plotting the 3D plot
weight_value = np.linspace(-2,2,50) # for plotting the graph between cost function vs weight vs bias, randomly generating the values of meshgrid
bias_value = np.linspace(-2,2,50)
W1,B1 = np.meshgrid(weight_value,bias_value) # plotting the meshgrid in 3D
#Plotting the graph of corresponding randomly generated weights and bias
cost_value = np.array([[cost_function(X_value,w_i,b_i,Y_true,Lambda) for w_i,b_i in zip(w_Row,b_Row)] for w_Row,b_Row in zip(W1,B1)])
```

```python
# Giving the path of directory for saving frames
os.makedirs('frames_Sub',exist_ok = True)
print(os.getcwd())
```

```python
# Implementing the gradient descent to minimize the loss function

for i in range(num_iter):
    # Computing gradient value with initial weight and bias
    sub_gradient_w,gradient_b = computing_gradient_cost(X_value,weight,bias,Y_true,Lambda)
    weight = weight-learning_rate*sub_gradient_w #Updating the weight
    bias = bias-learning_rate*gradient_b # Updating bias
    
    fig = plt.figure(figsize = (10,7)) # taking the figure size as defined in the function
    ax = fig.add_subplot(111,projection = '3d')
    # Plotting the surface area f cost function
    surf = ax.plot_surface(W1,B1,cost_value,cmap = 'viridis',edgecolor = 'none', alpha = 0.7)
    #Current position in the plot
    ax.scatter([weight],[bias],[cost_function(X_value,weight,bias,Y_true,Lambda)],color = 'r')
    ax.set_xlabel('weight(w)') # setting the x label
    ax.set_ylabel('Bias (b)') # setting the y label
    ax.set_zlabel('Cost function') # setting the z label
    ax.set_title(f'cost function in step {i+1}')
    ax.view_init(elev = 20, azim = 20)
    plt.savefig(f'frames_Sub/frames_{i:04d}.png')
    plt.close()
```
