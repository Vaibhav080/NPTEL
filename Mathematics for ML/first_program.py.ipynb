```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
class predictor:
    def __init__(self, num_inputs, learning_rate = 0.01, epochs=100):
        self.num_inputs=num_inputs
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.weights = np.random.rand(num_inputs)
        
    def predict(self,inputs):
        #calculate weghted sum
        weighted_sum= np.dot(inputs,self.weights)
        return weighted_sum
        
    def train(self,X,Y):
        Z1=[]
        Z2=[]
        for epoch in range(self.epochs):
            for i in range(len(X)):
                #Make prediction
                prediction = self.predict(X[i])
                #Calculate Error
                error = Y[i] - prediction
                #Update weights
                self.weights = self.weights + self.learning_rate*error*X[i]
                Z1.append(self.weights[0])
                Z2.append(self.weights[1])
                #Plot W1 vs W2
                plt.scatter(Z1,Z2,color='red',label='z1 vs z2')
                plt.title('weight 1 vs weight 2, LR = 0.01, epoch = 100')
                plt.xlabel('Weight 1')
                plt.ylabel('Weight 2')
                plt.scatter(self.weights[0],self.weights[1],color='g',label='w1 vs w2')
                print('Final weights', self.weights)
                plt.show()
```

```python
# Sample Dataset
X = np.array([[2,3],[4,5],[1,2],[3,4],[1,1],[2,2],[3,5],[1,5],[5,2]])
Y = np.array([13,23,8,18,5,10,21,17,16])
```

```python
# Initialize and train the perceptron without bias for regression
predictor = predictor(num_inputs=2)
print(predictor)
predictor.train(X,Y)

X_test = np.array([[4,3]])
Y_test = [[17]]

#Test the trained model

print("Testing the trained model...")
for i in range(len(X_test)):
    prediction = predictor.predict(X_test[i])
    print(f"Input: {X_test[i]}, Target: {Y_test[i]}, predicted_output: {prediction}")
```
