```python
import math
```

```python
def MLE_theta(sample_points):
    log_sum = 0
    for x in sample_points:
        log_sum += math.log(x)
    n = len(sample_points)
    return n/log_sum
```

```python
def main():
    sample_points = [2.5,3.1,4.0,2.8,3.7,8.9,4.3,10.3,1.1,5.6]
    theta = MLE_theta(sample_points)
    print(theta)
```

```python
if __name__ == "__main__":
    main()
```
