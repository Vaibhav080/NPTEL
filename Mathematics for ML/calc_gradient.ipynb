```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Define the range of x values so that the defined function can calculate the value
x_points = np.linspace(-2,2,50)

#Calculating the corresponding f(x) values using the function = 2e^(-x^2)

function = 2*np.exp(-x_points**2)
gradient_f = -4*x_points*np.exp(-x_points**2) ## Gradient of f(x)
```

```python
#Plotting the function

plt.plot(x_points,function)

plt.plot(x_points,gradient_f)

plt.xlabel('x')
plt.ylabel('$2e^{x^{2}}$')

plt.grid(True)
plt.show()
```
